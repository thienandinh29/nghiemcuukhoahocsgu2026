```python
import yfinance as yf
import pandas as pd

# 1. Download the data
ticker = "^GSPC"
start_date = "2010-01-01"
end_date = "2026-01-01"
sp500 = yf.download(ticker, start=start_date, end=end_date)
# Download data

# FIX: Flatten the columns if they have multiple levels
if isinstance(sp500.columns, pd.MultiIndex):
    sp500.columns = sp500.columns.get_level_values(0)

# Now this line will definitely work
sp500["Tomorrow"] = sp500["Close"].shift(-1)

# 2. Prepare the Target
# Create a column for tomorrow's closing price by shifting data back by 1 day
sp500["Tomorrow"] = sp500["Close"].shift(-1)

# Define the Target:
# Return 1 if Tomorrow's price is greater than Today's Close, else 0
sp500["Target"] = (sp500["Tomorrow"] > sp500["Close"]).astype(int)

# 3. Clean Up
# The shift(-1) creates a NaN (empty) value for the very last day (since there is no tomorrow)
# We must remove that row to avoid errors later
sp500 = sp500.dropna()

# Display the result to verify
print("Data with Target Column:")
# We select specific columns just to make it easy to read in the print statement
print(sp500[["Close", "Tomorrow", "Target"]].tail())

# 4. Save to CSV
sp500.to_csv("sp500_supervised_data.csv")
print("\nFile saved: sp500_supervised_data.csv")
```
