# 01 · Data Collection — S&P 500

**Goal.** Download daily S&P 500 (`^GSPC`) prices and build the supervised-learning label.

| Item | Value |
|---|---|
| Source | Yahoo Finance via `yfinance` |
| Period | 2010-01-01 → 2025-12-31 |
| Target | `1` if tomorrow's close > today's close, else `0` |
| Output | `sp500_supervised_data.csv` |

## 1. Setup

```python
pip install yfinance
```

## 2. Download and label

```python
import yfinance as yf
import pandas as pd

TICKER, START, END = "^GSPC", "2010-01-01", "2026-01-01"

# Download and flatten multi-level columns returned by yfinance
sp500 = yf.download(TICKER, start=START, end=END)
if isinstance(sp500.columns, pd.MultiIndex):
    sp500.columns = sp500.columns.get_level_values(0)

# Target: does the index close higher tomorrow than today?
sp500["Tomorrow"] = sp500["Close"].shift(-1)
sp500["Target"] = (sp500["Tomorrow"] > sp500["Close"]).astype(int)

# The last row has no "tomorrow" — drop it to avoid NaNs downstream
sp500 = sp500.dropna()

print("Data with Target column:")
print(sp500[["Close", "Tomorrow", "Target"]].tail())
```

## 3. Export

```python
OUTPUT = "sp500_supervised_data.csv"
sp500.to_csv(OUTPUT)
print(f"Saved {len(sp500):,} rows → {OUTPUT}")
```
